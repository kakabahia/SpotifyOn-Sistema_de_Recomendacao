```python
#importando os pacotes a serem utilizados
import pandas as pd
import numpy as np
```

```python
# Importar os arquivos com as músicas e ver as primeiras linhas
musicas = pd.read_csv("dadosspotify/Music Info.csv", low_memory=False)
musicas.head(3)
```

```python
# Importar o arquivo de quantidade de reproduções por usuário e ver as primeiras linhas
reproducoes = pd.read_csv("dadosspotify/User Listening History.csv", low_memory=False)
reproducoes.head(3)
```

```python
# Filtrando somente as colunas necessárias
# Seleciona somente as variaveis que iremos utilizar
musicas = musicas[["track_id", "name", "artist", "year"]]
musicas.rename (columns = {"track_id":"id_musica", "name":"nome_musica", "artist":"artista", "year":"ano"}, inplace = True)
musicas.head()
```

```python
# Filtrando somente as colunas necessárias
# Seleciona somente as variaveis que iremos utilizar
reproducoes.rename(columns={"track_id":"id_musica", "user_id":"id_usuario", "playcount":"qtd_reproducoes"}, inplace=True)
reproducoes.head()
```

```python
# Verificando a existência de valores nulos, não tem então não tem tratamento
musicas.isna().sum()
```

```python
# Verificando a existência de valores nulos, não tem então não tem tratamento
reproducoes.isna().sum()
```

```python
# Quantas vezes cada usuário aparece
reproducoes['id_usuario'].value_counts()
```

```python
# Cria um filtro para utilizarmos apenas usuarios que escutaram mais de 80 músicas
qt_reproducao = reproducoes['id_usuario'].value_counts() > 80
y = qt_reproducao[qt_reproducao].index
y.shape
```

```python
print(y)
reproducoes.shape
```

```python
# Filtro para atualizar o dataframe apenas com o que satisfazem a condição
reproducoes = reproducoes[reproducoes["id_usuario"].isin(y)]
reproducoes.shape
```

```python
# Mostra os tipos das variáveis do dataframe
musicas.info()
reproducoes.info()
# as colunas estão com mesmo tipo, então não precisa de tratamento
```

```python
reproducoes_e_musica = reproducoes.merge(musicas, on = "id_musica")
reproducoes_e_musica.head()
```

```python
reproducoes_e_musica.shape
```

```python
# Apagar duplicados de mesmo usuário para a mesma música
reproducoes_e_musica.drop_duplicates(["id_musica", "id_usuario"], inplace = True)
reproducoes_e_musica.shape
```

```python
# Apagar a variável id_filme pois não será utilizada
del reproducoes_e_musica['id_musica']
reproducoes_e_musica.head()
```

```python
# Quero que cada id_usuario seja uma variavel com o respectivo valor de nota 
# para cada filme avaliado
musicas_pivot = reproducoes_e_musica.pivot_table(columns = "id_usuario", index = "nome_musica", values = "qtd_reproducoes")
musicas_pivot.head(20)
```

```python
# Substitui os valores nulos por 0
musicas_pivot.fillna(0, inplace = True)
musicas_pivot.head(20)
```

```python
#importar csr_matriz do pacote SciPy
#possibilita a criação de matrizes sparsas
from scipy.sparse import csr_matrix

#Transformar o dataset em uma matriz sparsa
musicas_sparse = csr_matrix(musicas_pivot)
```

```python
type(musicas_sparse)
```

```python
#Importar o algoritmo KNN que calcula com base na distância
#do pacote SciKit Learn
from sklearn.neighbors import NearestNeighbors
```

```python
#Criando e treinando modelo preditivo
modelo = NearestNeighbors(algorithm="brute")
modelo.fit(musicas_sparse)
```

Vamos fazer previsões e sugestões de músicas

```python
#Wonderwall
distances, sugestions = modelo.kneighbors(musicas_pivot.filter(items = ["Wonderwall"], axis=0).values.reshape(1, -1))

for i in range(len(sugestions)):
    print(musicas_pivot.index[sugestions[i]])
```
